# Insurance Risk Analytics - EDA

This notebook contains exploratory data analysis for AlphaCare Insurance Solutions car insurance risk analytics challenge.

## 1. Import Libraries & Load Data
Import required libraries and load the insurance dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset (update filename as needed)
df = pd.read_csv('../data/insurance_claims.csv')
df.head()
```

## 2. Data Overview & Summary Statistics
Review the structure, data types, and summary statistics for key variables.

```python
# Data structure and types
df.info()

# Summary statistics for numerical columns
df.describe().T
```

## 3. Missing Value Analysis
Check for missing values and visualize their distribution.

```python
# Missing values count
df.isnull().sum().sort_values(ascending=False)

# Visualize missing values
plt.figure(figsize=(12,6))
sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
plt.title('Missing Values Heatmap')
plt.show()
```

## 4. Univariate Analysis
Explore distributions of key numerical and categorical variables.

```python
# Histogram for TotalPremium
plt.figure(figsize=(8,5))
sns.histplot(df['TotalPremium'], bins=30, kde=True)
plt.title('Distribution of TotalPremium')
plt.show()

# Histogram for TotalClaims
plt.figure(figsize=(8,5))
sns.histplot(df['TotalClaims'], bins=30, kde=True)
plt.title('Distribution of TotalClaims')
plt.show()

# Bar plot for Gender
plt.figure(figsize=(6,4))
sns.countplot(x='Gender', data=df)
plt.title('Distribution by Gender')
plt.show()
```

## 5. Bivariate & Multivariate Analysis
Explore relationships between key variables, including correlations and trends by geography and time.

```python
# Correlation matrix for numerical features
plt.figure(figsize=(12,8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()

# Scatter plot: TotalPremium vs TotalClaims by ZipCode
plt.figure(figsize=(10,6))
sns.scatterplot(x='TotalPremium', y='TotalClaims', hue='PostalCode', data=df, legend=False)
plt.title('TotalPremium vs TotalClaims by PostalCode')
plt.show()

# Grouped bar: Loss Ratio by Province
df['LossRatio'] = df['TotalClaims'] / df['TotalPremium']
province_loss = df.groupby('Province')['LossRatio'].mean().sort_values()
plt.figure(figsize=(10,5))
province_loss.plot(kind='bar', color='skyblue')
plt.ylabel('Average Loss Ratio')
plt.title('Average Loss Ratio by Province')
plt.show()
```
